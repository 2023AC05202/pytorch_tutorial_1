```python
import torch
```

```python
my_torch = torch.arange(10)
my_torch
```

```python
# reshape and view
my_torch = my_torch.reshape(2,5)
my_torch
```

```python
#reshape if we don't know the number of records
my_torch2 = torch.arange(10)
my_torch2
```

```python
my_torch2 = my_torch2.reshape(2,-1)
my_torch2
```

```python
#View
my_torch3 = torch.arange(10)
my_torch3
```

```python
my_torch3 = my_torch3.view(2,5)
my_torch3
```

```python
# reshape and view will update the origonal
my_torch4 = torch.arange(10)
my_torch4
```

```python
my_torch5 = my_torch4.reshape(2,5)
my_torch5
```

```python
my_torch4[1] = 4141
my_torch4
```

```python
my_torch5
```

```python
#Slice
my_torch7 = torch.arange(10)
my_torch7
```

```python
#Grab a specific item
my_torch7[6]
```

```python
# Grab a slice
my_torch8 = my_torch7.reshape(5,2)
my_torch8
```

```python
my_torch8[:,1]
```

```python
#return column
my_torch8[:,1:]
```
