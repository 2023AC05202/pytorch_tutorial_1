```python
import torch
```

```python
tensor_a = torch.tensor([1,2,3,4])
tensor_b = torch.tensor([5,6,7,8])
```

```python
#Addition
tensor_a + tensor_b
```

```python
#Long hand
torch.add(tensor_a, tensor_b)
```

```python
#Subtraction
tensor_b - tensor_a
```

```python
#Subtraction
torch.sub(tensor_b, tensor_a)
```

```python
#Multiplication
tensor_a * tensor_b
```

```python
#Multiplication function
torch.mul(tensor_a, tensor_b)
```

```python
#Division
tensor_b / tensor_a
```

```python
#Division Function
torch.div(tensor_b, tensor_a)
```

```python
#Remainder
tensor_b % tensor_a
```

```python
#Remainder function
torch.remainder(tensor_b, tensor_a)
```

```python
#Exponential
torch.pow(tensor_a, tensor_b)
```

```python
tensor_a.add_(tensor_b)
```
